
-   Code the solution to your assigned problem in Python (code chunk). Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Using that try to find the best time and space complexity solution!


````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    """
    Find the first value that appears more than once in the list.
    
    Args:
        nums: List of integers
        
    Returns:
        The first duplicate value, or -1 if no duplicates exist
    """
    seen = set()  # Keep track of numbers we've encountered
    
    for num in nums:
        if num in seen:  # O(1) lookup in set
            return num   # Found our first duplicate!
        seen.add(num)    # Remember this number for future checks
    
    return -1  # No duplicates found
````

````python
# Test the function with the provided examples
test_cases = [
    [3, 1, 4, 2, 5, 1, 6],  # Expected: 1
    [7, 8, 9, 10],           # Expected: -1
    [4, 5, 6, 4, 6],         # Expected: 4
    [2, 3, 2, 1, 5, 3],      # Expected: 2 (our custom example)
    [10, 20, 30, 40, 50]     # Expected: -1 (our custom example)
]

for i, test in enumerate(test_cases, 1):
    result = first_duplicate(test)
    print(f"Test {i}: {test} -> {result}")
````


-   Explain why your solution works


This solution works by using a set to keep track of numbers we've already seen as we iterate through the list. When we encounter a number that's already in our set, we immediately return it as the first duplicate. This approach is efficient because set lookups are O(1) operations, allowing us to detect duplicates in a single pass through the data.


-   Explain the problem’s time and space complexity


## Time and Space Complexity Explanation

**Time Complexity: O(n)**
- We iterate through the list exactly once, where n is the length of the input list
- For each element, we perform two O(1) operations: checking if it's in the set and adding it to the set
- Even in the worst case (no duplicates found), we visit every element once
- The set lookup and insertion operations are O(1) on average due to hash table implementation

**Space Complexity: O(n)**
- In the worst case scenario (when no duplicates exist), we store all n elements in the set
- In the best case scenario (first element is immediately duplicated), we only store 1 element: O(1)
- On average, we expect to store roughly n/2 elements before finding a duplicate
- The space grows linearly with input size, hence O(n)

**Why sets are optimal for this problem:**
- Sets provide O(1) average-case lookup time for membership testing
- Sets automatically handle uniqueness, which aligns perfectly with duplicate detection
- Alternative approaches like nested loops would be O(n²) time complexity

Explain the thinking to an alternative solution (no coding required, but a classmate reading this should be able to code it up based off your text)

Alternative Solution: Nested Loop Approach

An alternative approach would be to use a brute force method with nested loops. For each element in the list, check all subsequent elements to see if any match the current element. The outer loop would iterate through each position (i), and the inner loop would check positions (j) where j > i. When you find nums[i] == nums[j], return nums[i] as the first duplicate.

This approach has the advantage of using O(1) space complexity since no additional data structures are needed. However, it comes with a significant trade-off in time complexity, as it would be O(n²) in the worst case because every element might need to be compared with every other element. While this solution is simpler to understand and implement, it becomes impractical for large datasets due to its quadratic time complexity.
